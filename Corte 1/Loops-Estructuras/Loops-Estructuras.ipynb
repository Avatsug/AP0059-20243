factorial.py

```python
# calcula el factorial de un número
while True:
    # bucle infinito
    value = int(input("Enter a positive integer value: "))
    print("Value: ", value)
    a = isinstance(value, int) # qué es isinstance?
    if a == True and value > 0:
      # comprueba si es positivo
        fact = 1
        for i in range (1, value + 1):
            fact = fact*i
        print(f'The factorial of {value} is: ', fact)
    else:
        print("Please, enter a positive integer number")
```

primos.py

```python
import time
inicio = time.time()

# determina si el número del rango es primo o no, si lo es, lo imprime
for i in range(0,31):
    conta = 0
    for n in range(1, i+1):
        residue = i%n
        if residue == 0:
            conta = conta + 1

    if conta == 2:
        print(f'{i} es un primo')

fin = time.time()
print("t = ", (fin - inicio)*1000) # tiempo que tarda en encontrar los números primos
```
