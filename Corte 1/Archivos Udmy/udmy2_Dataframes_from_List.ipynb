```python
# Crear un DataFrame

import pandas as pd # Se importa la biblioteca 'pandas' y se le asigna un nombre para facilitar su uso 'pd'
```

```python
numbers = [1, 2, 3, 4, 5, 6] # Se crea un arreglo de datos para usarlo en el DataFrame
```

```python
pd.DataFrame(numbers) # Se le asigna
```

```python
data = [["Jhon",  23], ["Mary", 18], ["Jhon", 31]] # 'data' es una Lista de Listas
```

```python
pd.DataFrame(data) # Dataframe donde la info son las listas de 'data'
```

```python
pd.DataFrame(data, columns = ["name", "age"]) # Cambia el nombre de las columnas
```

```python
pd.DataFrame(data, columns = ["name", "age"], index = ["id1", "id2", "id3"]) # Cambia el nombre del índice
```
