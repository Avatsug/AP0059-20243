```python
import random
import csv
import pandas as pd

# generar el archivo .csv
with open('numeros.csv', mode='w', newline='', encoding='utf-8') as archivo:
    escritor = csv.writer(archivo)
    escritor.writerow(['Número'])

    # bucle while para escribir cada dato en el archivo .csv
    j=0
    while j<10:
        # generar número pseudo-aleatorio entre 0 y 1
        n = str(random.random())
        j+=1
        escritor.writerow([n])

# crear dataframe
df = pd.read_csv('numeros.csv', delimiter=',')

# empezar las filas desde 1
# df.index = range(1, len(df) + 1)

print(df)
```
